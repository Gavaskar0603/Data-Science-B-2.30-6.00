```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as ss
```

```python
data = np.random.normal(loc=0,scale=1,size=500)
data
```

```python
ss.probplot(data,dist="norm",plot=plt)
plt.title("Normal QQ-plot")
plt.xlabel("Quantile")
plt.ylabel("Order Value")
plt.show()
```

```python
data1 = np.random.exponential(scale=1,size=500)
data1
```

```python
ss.probplot(data1,dist="expon",plot=plt)
plt.title("Expon QQ-plot")
plt.xlabel("Quantile")
plt.ylabel("Order Value")
plt.show()
```

```python
data1 = np.random.uniform(size=500)
data1
```

```python
ss.probplot(data1,dist="uniform",plot=plt)
plt.title("Expon QQ-plot")
plt.xlabel("Quantile")
plt.ylabel("Order Value")
plt.show()
```
