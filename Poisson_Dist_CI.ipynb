```python
from scipy.stats import binom as b
import numpy as np
import matplotlib.pyplot as plt
```

```python
n=10
p=0.6
x=np.arange(0,n+1)
pmf = b.pmf(x,n,p)
pmf
```

```python
plt.figure(figsize=(8,5))
plt.bar(x,pmf)
plt.xlabel("Number of Success")
plt.ylabel("Probability percent")
plt.title("Binomial Distribution")
plt.show()
```

```python
cdf = b.cdf(x,n,p)
cdf
```

```python
plt.figure(figsize=(8,5))
plt.plot(x,cdf)
plt.xlabel("Number of Success")
plt.ylabel("Probability percent")
plt.title("Binomial Distribution")
plt.show()
```

```python
from scipy import stats

salary = [18000,25000,35000,48000,52000,150000,60000,55000,78000]
sample_mean = np.mean(salary) 
CI = stats.t.interval(
    confidence=0.95,
    df=len(salary)-1,
    loc= sample_mean,
    scale=stats.sem(salary)
)

print(f"This is my Confidence level {CI}")
```
